```python

import tensorflow as tf
import pandas as pd
from tensorflow import keras
import tensorflow_datasets as tfds
import numpy as np
import matplotlib.pyplot as plt

print(tf.__version__)
```

```python
# get data files
!wget https://cdn.freecodecamp.org/project-data/sms/train-data.tsv
!wget https://cdn.freecodecamp.org/project-data/sms/valid-data.tsv

train_file_path = "train-data.tsv"
test_file_path = "valid-data.tsv"
```

```python
train_df = pd.read_csv(train_file_path, sep='\t', header=None, names=['label', 'message'])
test_df = pd.read_csv(test_file_path, sep='\t', header=None, names=['label', 'message'])
```

```python
print('Training data head:')
display(train_df.head())
```

```python
print('Test data head:')
display(test_df.head())
```

```python
train_df['label'] = train_df['label'].map({'ham': 0, 'spam': 1})
test_df['label'] = test_df['label'].map({'ham': 0, 'spam': 1})

display(train_df.head())
```

```python

display(test_df.head())
```

```python

train_df['label'].value_counts()
```

```python
from tensorflow.keras.layers import TextVectorization

vocab_size = 2000
max_len = 50

vectorizer = TextVectorization(
    max_tokens = vocab_size,
    output_mode = 'int',
    output_sequence_length = max_len
)
```

```python
train_df.head()
```

```python
vectorizer.adapt(train_df['message'])
X_train = vectorizer(train_df['message'])
Y_train = train_df['label']
X_test = vectorizer(test_df['message'])
Y_test = test_df['label']
```

```python
X_train[0]
```

```python
from tensorflow.keras.layers import TextVectorization
temp = TextVectorization()
temp.adapt(train_df['message'])
print(len(temp.get_vocabulary()))
```

```python
from collections import Counter

all_words = []
for message in train_df['message']:
    words = message.lower().split()
    all_words.extend(words)

word_counts = Counter(all_words)

print("total unique words:", len(word_counts))
print("words appearing >=2 times:", sum(1 for c in word_counts.values() if c >= 2))
print("words appearing >=5 times:", sum(1 for c in word_counts.values() if c >= 5))
print("words appearing >=10 times:", sum(1 for c in word_counts.values() if c >= 10))

print(word_counts.most_common(10))
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

Embedding_dim = 16

model = Sequential([
    Embedding(input_dim = vocab_size, output_dim = Embedding_dim),
    LSTM(32),
    Dense(1, activation = 'sigmoid')
 ])

model.compile(
    loss = 'binary_crossentropy',
    optimizer = 'adam',
    metrics = ['accuracy']
)

model(X_train[:1])

model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)


history = model.fit(
    X_train, Y_train,
    validation_data = (X_test, Y_test),
    epochs = 30,
    callbacks = [early_stop],
    verbose = 2
)
```

```python
loss, acc = model.evaluate(X_test, Y_test, verbose=0)
print(loss, acc)
```

```python

def predict_message(pred_text):
  prediction = []
  vectorized = vectorizer([pred_text])
  prediction = model.predict(vectorized)
  probability = float(prediction[0][0])
  label = "spam" if probability >= 0.5 else "ham"
  prediction = [probability, label]

  return (prediction)

pred_text = "how are you doing today?"

prediction = predict_message(pred_text)
print(prediction)
```
